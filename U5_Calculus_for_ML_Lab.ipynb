```python
import numpy as np
import sympy as sp

def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 using finite difference
x = 2
h = 1e-6

numerical_derivative = (g(x + h) - g(x)) / h
print("Numerical derivative:", numerical_derivative)

# 2. Symbolic derivative
x_sym = sp.symbols('x')
g_sym = x_sym**3 + 2*x_sym

symbolic_derivative = sp.diff(g_sym, x_sym)
print("Symbolic derivative:", symbolic_derivative)

# 3. Evaluate symbolic derivative at x = 2
symbolic_value = symbolic_derivative.subs(x_sym, 2)
print("Symbolic derivative at x = 2:", symbolic_value)

# Compare
print("Match:", np.isclose(numerical_derivative, float(symbolic_value)))
```

```python


x, y = sp.symbols('x y')

f2 = x**2 + 3*x*y + y**2

print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

h2 = x**2 * y + sp.sin(y)

# 1. Compute dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient
gradient = [dh_dx, dh_dy]
print("Gradient =", gradient)

# 3. Evaluate gradient at (x=2, y=0)
gradient_at_point = [g.subs({x: 2, y: 0}) for g in gradient]
print("Gradient at (2, 0) =", gradient_at_point)
```

```python
import sympy as sp

# Define x
x = sp.symbols('x')

# y = exp(x**2 + 1)
y = sp.exp(x**2 + 1)

# 1. By hand: outer derivative × inner derivative
by_hand = sp.exp(x**2 + 1) * (2*x)

# 2. With sp.diff
by_sympy = sp.diff(y, x)

print("By hand:", by_hand)
print("With sp.diff:", by_sympy)

# 3. Confirm they match
print("Do they match?", sp.simplify(by_hand - by_sympy) == 0)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)

Wa = np.random.randn(4, 8) * 0.1   # input -> hidden
Wb = np.random.randn(8, 1) * 0.1   # hidden -> output

# ReLU function
def relu(x):
    return np.maximum(0, x)

# 1. Forward pass
z1 = Xb @ Wa
h = relu(z1)
y_hat = h @ Wb

loss = np.mean((y_hat - Yb)**2)

print("Loss before:", loss)

# 2. Backward pass
# d(loss)/d(y_hat)
dy = (2 / len(Yb)) * (y_hat - Yb)

# Gradient for Wb
dWb = h.T @ dy

# Gradient flowing back into hidden layer
dh = dy @ Wb.T

# Gradient through ReLU
dz1 = dh * (z1 > 0)

# Gradient for Wa
dWa = Xb.T @ dz1

# 3. One gradient-descent step
lr = 0.05

Wa = Wa - lr * dWa
Wb = Wb - lr * dWb

# Forward pass again to calculate new loss
z1 = Xb @ Wa
h = relu(z1)
y_hat = h @ Wb

loss_after = np.mean((y_hat - Yb)**2)

print("Loss after:", loss_after)
```

```python
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

# 1. Hessian of f = x**4 + y**2
f = x**4 + y**2
H = sp.hessian(f, (x, y))

print("Hessian of f:")
sp.pprint(H)

# 2. Gradient descent to minimise (x - 7)**2
xv = 0
lr = 0.1

for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad

print("Final x:", xv)
```
